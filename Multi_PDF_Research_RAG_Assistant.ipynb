```python
import fitz
import os
import re
from tqdm.auto import tqdm

def clean_text(text):
    """Clean extracted PDF text."""

    # Replace null characters
    text = text.replace("\x00", " ")

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def extract_pdf(pdf_path):
    """Extract text page-by-page from a PDF."""

    document = fitz.open(pdf_path)

    pages = []

    for page_number, page in enumerate(document, start=1):

        text = page.get_text("text")

        text = clean_text(text)

        if text:
            pages.append({
                "document": os.path.basename(pdf_path),
                "path": pdf_path,
                "page": page_number,
                "text": text
            })

    document.close()

    return pages
```

```python
import fitz
import os
import re
from tqdm.auto import tqdm

def clean_text(text):
    text = text.replace("\x00", " ")
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def extract_pdf(pdf_path):

    pdf_path = os.path.abspath(pdf_path)

    if not os.path.isfile(pdf_path):
        raise FileNotFoundError(f"PDF not found: {pdf_path}")

    doc = fitz.open(pdf_path)

    pages = []

    for page_number in range(len(doc)):

        page = doc[page_number]

        text = page.get_text("text")
        text = clean_text(text)

        if text:
            pages.append({
                "document": os.path.basename(pdf_path),
                "path": pdf_path,
                "page": page_number + 1,
                "text": text
            })

    doc.close()

    return pages
```

```python
def chunk_text(text, chunk_size=600, overlap=100):

    words = text.split()

    chunks = []

    start = 0

    while start < len(words):

        end = min(start + chunk_size, len(words))

        chunk = " ".join(words[start:end])

        if chunk.strip():
            chunks.append(chunk)

        if end >= len(words):
            break

        start = end - overlap

    return chunks
```

```python
chunks = []

chunk_id = 0

for page in tqdm(all_pages, desc="Creating chunks"):

    page_chunks = chunk_text(
        page["text"],
        chunk_size=600,
        overlap=100
    )

    for chunk_number, chunk in enumerate(page_chunks):

        chunks.append({
            "chunk_id": chunk_id,
            "document": page["document"],
            "path": page["path"],
            "page": page["page"],
            "chunk_on_page": chunk_number,
            "text": chunk
        })

        chunk_id += 1

print("\n==============================")
print("CHUNKING COMPLETE")
print("==============================")
print("Total chunks:", len(chunks))
```

```python
def reciprocal_rank_fusion(
    dense_indices,
    bm25_indices,
    k=60
):

    fused_scores = {}

    for rank, idx in enumerate(
        dense_indices,
        start=1
    ):

        idx = int(idx)

        fused_scores[idx] = (
            fused_scores.get(idx, 0)
            + 1.0 / (k + rank)
        )

    for rank, idx in enumerate(
        bm25_indices,
        start=1
    ):

        idx = int(idx)

        fused_scores[idx] = (
            fused_scores.get(idx, 0)
            + 1.0 / (k + rank)
        )

    ranked = sorted(
        fused_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return ranked
```

```python
def retrieve_context(
    query,
    dense_k=50,
    bm25_k=50,
    rerank_k=30,
    final_k=8
):

    # ==========================================
    # HYBRID RETRIEVAL
    # ==========================================

    retrieval_results = hybrid_retrieve(
        query,
        dense_k=dense_k,
        bm25_k=bm25_k
    )

    # ==========================================
    # RRF
    # ==========================================

    rrf_results = reciprocal_rank_fusion(
        retrieval_results["dense_indices"],
        retrieval_results["bm25_indices"]
    )

    candidates = rrf_results[:rerank_k]

    # ==========================================
    # RERANK
    # ==========================================

    pairs = []

    for idx, rrf_score in candidates:

        idx = int(idx)

        text = chunks_df.iloc[idx]["text"]

        pairs.append([
            query,
            text
        ])

    scores = reranker.predict(
        pairs,
        batch_size=16,
        show_progress_bar=False
    )

    # ==========================================
    # COMBINE
    # ==========================================

    reranked = []

    for (idx, rrf_score), score in zip(
        candidates,
        scores
    ):

        reranked.append({
            "idx": int(idx),
            "rrf_score": float(rrf_score),
            "rerank_score": float(score)
        })

    # Highest relevance first
    reranked.sort(
        key=lambda x: x["rerank_score"],
        reverse=True
    )

    return reranked[:final_k]
```

```python
def filter_reranked_results(
    results,
    min_score=-5.0,
    max_results=8
):
    """
    Keep only sufficiently relevant reranked chunks.

    Note:
    Reranker score ranges depend on the model and
    should be calibrated using your own documents.
    """

    filtered = [
        result
        for result in results
        if result["rerank_score"] >= min_score
    ]

    return filtered[:max_results]
```

```python
def remove_duplicate_sources(
    results,
    chunks_df,
    max_results=8
):

    selected = []

    seen = set()

    for result in results:

        idx = result["idx"]

        row = chunks_df.iloc[idx]

        # Document + page + chunk
        key = (
            row["document"],
            row["page"],
            row["chunk_on_page"]
        )

        if key in seen:
            continue

        seen.add(key)

        selected.append(result)

        if len(selected) >= max_results:
            break

    return selected
```

```python
def select_context_chunks(
    results,
    chunks_df,
    max_chunks=8
):

    # Remove duplicates first
    results = remove_duplicate_sources(
        results,
        chunks_df,
        max_results=max_chunks * 2
    )

    # Apply relevance filtering
    results = filter_reranked_results(
        results,
        min_score=-5.0,
        max_results=max_chunks
    )

    return results
```

```python
def format_sources(sources):

    if not sources:
        return "No sources found."

    output = "### Sources\n\n"

    seen = set()

    for i, source in enumerate(sources, 1):

        row = chunks_df.iloc[source["idx"]]

        document = row["document"]
        page = row["page"]

        key = (document, page)

        if key in seen:
            continue

        seen.add(key)

        output += (
            f"**{len(seen)}. {document}**  \n"
            f"Page: **{page}**  \n"
            f"Relevance score: "
            f"`{source['rerank_score']:.4f}`\n\n"
        )

    return output
```

```python
# ============================================================
# STEP 2 — COMPREHENSIVE MULTI-PDF RETRIEVER
# ============================================================

def dense_retrieve(
    query,
    top_k=150
):
    """
    Retrieve a large candidate pool using BGE-M3 + FAISS.
    """

    query_embedding = embedding_model.encode(
        [query],
        batch_size=1,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False,
        device="cuda"
    ).astype("float32")

    scores, indices = gpu_index.search(
        query_embedding,
        min(top_k, len(chunks_df))
    )

    return scores[0], indices[0]


def bm25_retrieve(
    query,
    top_k=150
):
    """
    Retrieve a large candidate pool using BM25.
    """

    query_tokens = bm25_tokenize(query)

    scores = bm25.get_scores(query_tokens)

    top_k = min(top_k, len(scores))

    indices = np.argsort(scores)[::-1][:top_k]

    return scores[indices], indices


def reciprocal_rank_fusion(
    dense_indices,
    bm25_indices,
    k=60
):
    """
    Combine dense and lexical retrieval using RRF.
    """

    fused_scores = {}

    # Dense ranking
    for rank, idx in enumerate(dense_indices, start=1):

        idx = int(idx)

        fused_scores[idx] = (
            fused_scores.get(idx, 0)
            + 1.0 / (k + rank)
        )

    # BM25 ranking
    for rank, idx in enumerate(bm25_indices, start=1):

        idx = int(idx)

        fused_scores[idx] = (
            fused_scores.get(idx, 0)
            + 1.0 / (k + rank)
        )

    ranked = sorted(
        fused_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return ranked


def comprehensive_retrieve(
    query,
    dense_k=150,
    bm25_k=150,
    rerank_k=100
):
    """
    Retrieve a large set of candidates across ALL PDFs.

    Pipeline:

    Query
      ↓
    FAISS
      +
    BM25
      ↓
    RRF
      ↓
    Large candidate pool
      ↓
    Reranker
    """

    # --------------------------------------------------------
    # Dense retrieval
    # --------------------------------------------------------

    dense_scores, dense_indices = dense_retrieve(
        query,
        top_k=dense_k
    )

    # --------------------------------------------------------
    # BM25 retrieval
    # --------------------------------------------------------

    bm25_scores, bm25_indices = bm25_retrieve(
        query,
        top_k=bm25_k
    )

    # --------------------------------------------------------
    # RRF
    # --------------------------------------------------------

    rrf_results = reciprocal_rank_fusion(
        dense_indices,
        bm25_indices
    )

    candidates = rrf_results[:rerank_k]

    # --------------------------------------------------------
    # Build reranker pairs
    # --------------------------------------------------------

    pairs = []

    for idx, rrf_score in candidates:

        idx = int(idx)

        text = chunks_df.iloc[idx]["text"]

        pairs.append([
            query,
            text
        ])

    # --------------------------------------------------------
    # Cross encoder reranking
    # --------------------------------------------------------

    rerank_scores = reranker.predict(
        pairs,
        batch_size=16,
        show_progress_bar=False
    )

    results = []

    for (idx, rrf_score), score in zip(
        candidates,
        rerank_scores
    ):

        idx = int(idx)

        row = chunks_df.iloc[idx]

        results.append({

            "idx": idx,

            "document": row["document"],

            "page": int(row["page"]),

            "chunk_on_page": int(row["chunk_on_page"]),

            "rrf_score": float(rrf_score),

            "rerank_score": float(score),

            "text": row["text"]

        })

    # --------------------------------------------------------
    # Sort by reranker score
    # --------------------------------------------------------

    results.sort(
        key=lambda x: x["rerank_score"],
        reverse=True
    )

    return results
```

```python
# ============================================================
# STEP 2 — COMPREHENSIVE MULTI-PDF RETRIEVER
# ============================================================

def dense_retrieve(
    query,
    top_k=150
):
    """
    Retrieve a large candidate pool using BGE-M3 + FAISS.
    """

    query_embedding = embedding_model.encode(
        [query],
        batch_size=1,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False,
        device="cuda"
    ).astype("float32")

    scores, indices = gpu_index.search(
        query_embedding,
        min(top_k, len(chunks_df))
    )

    return scores[0], indices[0]


def bm25_retrieve(
    query,
    top_k=150
):
    """
    Retrieve a large candidate pool using BM25.
    """

    query_tokens = bm25_tokenize(query)

    scores = bm25.get_scores(query_tokens)

    top_k = min(top_k, len(scores))

    indices = np.argsort(scores)[::-1][:top_k]

    return scores[indices], indices


def reciprocal_rank_fusion(
    dense_indices,
    bm25_indices,
    k=60
):
    """
    Combine dense and lexical retrieval using RRF.
    """

    fused_scores = {}

    # Dense ranking
    for rank, idx in enumerate(dense_indices, start=1):

        idx = int(idx)

        fused_scores[idx] = (
            fused_scores.get(idx, 0)
            + 1.0 / (k + rank)
        )

    # BM25 ranking
    for rank, idx in enumerate(bm25_indices, start=1):

        idx = int(idx)

        fused_scores[idx] = (
            fused_scores.get(idx, 0)
            + 1.0 / (k + rank)
        )

    ranked = sorted(
        fused_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return ranked


def comprehensive_retrieve(
    query,
    dense_k=150,
    bm25_k=150,
    rerank_k=100
):
    """
    Retrieve a large set of candidates across ALL PDFs.

    Pipeline:

    Query
      ↓
    FAISS
      +
    BM25
      ↓
    RRF
      ↓
    Large candidate pool
      ↓
    Reranker
    """

    # --------------------------------------------------------
    # Dense retrieval
    # --------------------------------------------------------

    dense_scores, dense_indices = dense_retrieve(
        query,
        top_k=dense_k
    )

    # --------------------------------------------------------
    # BM25 retrieval
    # --------------------------------------------------------

    bm25_scores, bm25_indices = bm25_retrieve(
        query,
        top_k=bm25_k
    )

    # --------------------------------------------------------
    # RRF
    # --------------------------------------------------------

    rrf_results = reciprocal_rank_fusion(
        dense_indices,
        bm25_indices
    )

    candidates = rrf_results[:rerank_k]

    # --------------------------------------------------------
    # Build reranker pairs
    # --------------------------------------------------------

    pairs = []

    for idx, rrf_score in candidates:

        idx = int(idx)

        text = chunks_df.iloc[idx]["text"]

        pairs.append([
            query,
            text
        ])

    # --------------------------------------------------------
    # Cross encoder reranking
    # --------------------------------------------------------

    rerank_scores = reranker.predict(
        pairs,
        batch_size=16,
        show_progress_bar=False
    )

    results = []

    for (idx, rrf_score), score in zip(
        candidates,
        rerank_scores
    ):

        idx = int(idx)

        row = chunks_df.iloc[idx]

        results.append({

            "idx": idx,

            "document": row["document"],

            "page": int(row["page"]),

            "chunk_on_page": int(row["chunk_on_page"]),

            "rrf_score": float(rrf_score),

            "rerank_score": float(score),

            "text": row["text"]

        })

    # --------------------------------------------------------
    # Sort by reranker score
    # --------------------------------------------------------

    results.sort(
        key=lambda x: x["rerank_score"],
        reverse=True
    )

    return results
```

```python
# ============================================================
# STEP 5 — DOCUMENT-DIVERSE SELECTION
# ============================================================

def select_diverse_results(
    results,
    max_results=30,
    max_per_document=5
):
    """
    Select relevant results while preventing one PDF
    from dominating the context.
    """

    selected = []

    document_counts = {}

    for result in results:

        document = result["document"]

        current_count = document_counts.get(
            document,
            0
        )

        if current_count >= max_per_document:
            continue

        selected.append(result)

        document_counts[document] = (
            current_count + 1
        )

        if len(selected) >= max_results:
            break

    return selected
```

```python
# ============================================================
# STEP 6 — BUILD COMPREHENSIVE CONTEXT
# ============================================================

def build_comprehensive_context(
    results,
    max_chars_per_chunk=6000
):

    context_parts = []

    for i, result in enumerate(results, start=1):

        text = result["text"]

        if len(text) > max_chars_per_chunk:
            text = text[:max_chars_per_chunk]

        source = (
            f"[SOURCE {i}]\n"
            f"Document: {result['document']}\n"
            f"Page: {result['page']}\n"
            f"Chunk: {result['chunk_on_page']}\n"
            f"Relevance: {result['rerank_score']:.4f}\n\n"
            f"{text}\n"
            f"[/SOURCE {i}]"
        )

        context_parts.append(source)

    return "\n\n".join(context_parts)
```

```python
# ============================================================
# STEP 4 — SEARCH ALL PDF DOCUMENTS
# ============================================================

def search_all_documents(
    query,
    semantic_k=None,
    lexical_k=None
):
    """
    Search the entire PDF collection.

    Every PDF gets evaluated through its document-level
    representation and BM25 score.
    """

    num_documents = len(documents_df)

    # --------------------------------------------------------
    # Semantic search
    # --------------------------------------------------------

    query_embedding = embedding_model.encode(
        [query],
        batch_size=1,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False,
        device="cuda"
    ).astype("float32")

    semantic_scores, semantic_indices = (
        document_gpu_index.search(
            query_embedding,
            num_documents
        )
    )

    semantic_scores = semantic_scores[0]
    semantic_indices = semantic_indices[0]

    # --------------------------------------------------------
    # BM25 search
    # --------------------------------------------------------

    query_tokens = bm25_tokenize(query)

    lexical_scores = document_bm25.get_scores(
        query_tokens
    )

    # --------------------------------------------------------
    # Normalize scores
    # --------------------------------------------------------

    def minmax_normalize(scores):

        scores = np.asarray(scores, dtype=np.float32)

        min_score = scores.min()
        max_score = scores.max()

        if max_score - min_score < 1e-8:
            return np.zeros_like(scores)

        return (
            (scores - min_score)
            / (max_score - min_score)
        )

    semantic_normalized = minmax_normalize(
        semantic_scores
    )

    lexical_normalized = minmax_normalize(
        lexical_scores
    )

    # --------------------------------------------------------
    # Combine document scores
    # --------------------------------------------------------

    combined_scores = (
        0.65 * semantic_normalized
        +
        0.35 * lexical_normalized
    )

    ranked_indices = np.argsort(
        combined_scores
    )[::-1]

    # --------------------------------------------------------
    # Return ALL documents
    # --------------------------------------------------------

    results = []

    for rank, idx in enumerate(
        ranked_indices,
        start=1
    ):

        idx = int(idx)

        results.append({

            "document_idx": idx,

            "document": documents_df.iloc[idx]["document"],

            "path": documents_df.iloc[idx]["path"],

            "pages": int(
                documents_df.iloc[idx]["num_pages"]
            ),

            "chunks": int(
                documents_df.iloc[idx]["num_chunks"]
            ),

            "semantic_score": float(
                semantic_normalized[idx]
            ),

            "bm25_score": float(
                lexical_normalized[idx]
            ),

            "document_score": float(
                combined_scores[idx]
            ),

            "rank": rank

        })

    return results
```

```python
# ============================================================
# STEP 7 — MULTI-PDF CHUNK RETRIEVAL
# ============================================================

def retrieve_chunks_across_documents(
    query,
    max_documents=100,
    chunks_per_document=5,
    rerank_candidates=150
):
    """
    Search across multiple relevant PDFs and collect
    relevant chunks from each document.
    """

    # --------------------------------------------------------
    # 1. Search ALL PDFs
    # --------------------------------------------------------

    document_results = search_all_documents(query)

    # Take the highest-ranked documents for chunk retrieval.
    # IMPORTANT: all PDFs were already searched above.
    selected_documents = document_results[:max_documents]

    selected_document_names = {
        r["document"]
        for r in selected_documents
    }

    print(
        f"All PDFs searched: "
        f"{len(document_results):,}"
    )

    print(
        f"PDFs selected for detailed chunk search: "
        f"{len(selected_document_names):,}"
    )

    # --------------------------------------------------------
    # 2. Get chunks belonging to selected PDFs
    # --------------------------------------------------------

    document_mask = chunks_df["document"].isin(
        selected_document_names
    )

    candidate_chunk_indices = np.where(
        document_mask.values
    )[0]

    print(
        f"Candidate chunks from selected PDFs: "
        f"{len(candidate_chunk_indices):,}"
    )

    # --------------------------------------------------------
    # 3. Semantic search among candidate chunks
    # --------------------------------------------------------

    query_embedding = embedding_model.encode(
        [query],
        batch_size=1,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False,
        device="cuda"
    ).astype("float32")

    candidate_embeddings = embeddings[
        candidate_chunk_indices
    ].astype("float32")

    # Cosine similarity because embeddings are normalized
    semantic_scores = (
        candidate_embeddings @ query_embedding[0]
    )

    semantic_order = np.argsort(
        semantic_scores
    )[::-1]

    # --------------------------------------------------------
    # 4. BM25 within candidate chunks
    # --------------------------------------------------------

    query_tokens = bm25_tokenize(query)

    bm25_scores_all = bm25.get_scores(
        query_tokens
    )

    candidate_bm25_scores = bm25_scores_all[
        candidate_chunk_indices
    ]

    bm25_order = np.argsort(
        candidate_bm25_scores
    )[::-1]

    # --------------------------------------------------------
    # 5. RRF
    # --------------------------------------------------------

    rrf_scores = {}

    # Semantic ranking
    for rank, position in enumerate(
        semantic_order,
        start=1
    ):

        chunk_idx = int(
            candidate_chunk_indices[position]
        )

        rrf_scores[chunk_idx] = (
            rrf_scores.get(chunk_idx, 0)
            + 1.0 / (60 + rank)
        )

    # BM25 ranking
    for rank, position in enumerate(
        bm25_order,
        start=1
    ):

        chunk_idx = int(
            candidate_chunk_indices[position]
        )

        rrf_scores[chunk_idx] = (
            rrf_scores.get(chunk_idx, 0)
            + 1.0 / (60 + rank)
        )

    rrf_ranked = sorted(
        rrf_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    # --------------------------------------------------------
    # 6. Keep enough candidates for reranking
    # --------------------------------------------------------

    rrf_candidates = rrf_ranked[
        :rerank_candidates
    ]

    print(
        f"RRF candidates: "
        f"{len(rrf_candidates):,}"
    )

    # --------------------------------------------------------
    # 7. Reranker
    # --------------------------------------------------------

    pairs = []

    for chunk_idx, rrf_score in rrf_candidates:

        text = chunks_df.iloc[
            int(chunk_idx)
        ]["text"]

        pairs.append([
            query,
            text
        ])

    rerank_scores = reranker.predict(
        pairs,
        batch_size=16,
        show_progress_bar=True
    )

    reranked = []

    for (chunk_idx, rrf_score), score in zip(
        rrf_candidates,
        rerank_scores
    ):

        chunk_idx = int(chunk_idx)

        row = chunks_df.iloc[chunk_idx]

        reranked.append({

            "idx": chunk_idx,

            "document": row["document"],

            "page": int(row["page"]),

            "chunk_on_page": int(
                row["chunk_on_page"]
            ),

            "rrf_score": float(rrf_score),

            "rerank_score": float(score),

            "text": row["text"]

        })

    # Highest reranker score first
    reranked.sort(
        key=lambda x: x["rerank_score"],
        reverse=True
    )

    # --------------------------------------------------------
    # 8. Select multiple chunks per document
    # --------------------------------------------------------

    final_results = []

    document_chunk_counts = {}

    for result in reranked:

        document = result["document"]

        current_count = document_chunk_counts.get(
            document,
            0
        )

        if current_count >= chunks_per_document:
            continue

        final_results.append(result)

        document_chunk_counts[document] = (
            current_count + 1
        )

    return final_results
```

```python
# ============================================================
# STEP 9 — EXHAUSTIVE ALL-PDF CHUNK RETRIEVAL
# ============================================================

def exhaustive_multi_pdf_retrieve(
    query,
    chunks_per_pdf=5,
    rerank_batch_size=32
):
    """
    Exhaustive multi-PDF retrieval.

    IMPORTANT:
    Every chunk in every indexed PDF is considered.

    Pipeline:

        ALL CHUNKS
           ↓
      BGE-M3 semantic
           +
         BM25
           ↓
    Select best chunks PER PDF
           ↓
        Reranker
           ↓
    Relevant evidence across PDFs
    """

    total_chunks = len(chunks_df)
    total_documents = chunks_df["document"].nunique()

    print("=" * 80)
    print("EXHAUSTIVE MULTI-PDF RETRIEVAL")
    print("=" * 80)

    print(f"PDFs:   {total_documents:,}")
    print(f"Chunks: {total_chunks:,}")

    # --------------------------------------------------------
    # 1. Encode query
    # --------------------------------------------------------

    query_embedding = embedding_model.encode(
        [query],
        batch_size=1,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False,
        device="cuda"
    ).astype("float32")[0]

    # --------------------------------------------------------
    # 2. Semantic similarity against EVERY chunk
    # --------------------------------------------------------

    print("\nSearching every chunk with BGE-M3...")

    # Because embeddings are normalized,
    # dot product = cosine similarity.
    semantic_scores = embeddings.astype(
        "float32"
    ) @ query_embedding

    print(
        f"✓ Semantic scores calculated for "
        f"{len(semantic_scores):,} chunks"
    )

    # --------------------------------------------------------
    # 3. BM25 score for EVERY chunk
    # --------------------------------------------------------

    print("Searching every chunk with BM25...")

    query_tokens = bm25_tokenize(query)

    bm25_scores = bm25.get_scores(
        query_tokens
    )

    print(
        f"✓ BM25 scores calculated for "
        f"{len(bm25_scores):,} chunks"
    )

    # --------------------------------------------------------
    # 4. Normalize scores
    # --------------------------------------------------------

    def normalize_array(values):

        values = np.asarray(
            values,
            dtype=np.float32
        )

        min_value = values.min()
        max_value = values.max()

        if max_value - min_value < 1e-8:
            return np.zeros_like(values)

        return (
            (values - min_value)
            / (max_value - min_value)
        )

    semantic_norm = normalize_array(
        semantic_scores
    )

    bm25_norm = normalize_array(
        bm25_scores
    )

    # --------------------------------------------------------
    # 5. Hybrid score
    # --------------------------------------------------------

    hybrid_scores = (
        0.65 * semantic_norm
        +
        0.35 * bm25_norm
    )

    # --------------------------------------------------------
    # 6. Select best chunks PER PDF
    #
    # This is the important part.
    #
    # A PDF cannot disappear just because another PDF has
    # more highly-ranked chunks.
    # --------------------------------------------------------

    print(
        f"\nSelecting top {chunks_per_pdf} "
        f"candidate chunks from EACH PDF..."
    )

    candidate_indices = []

    document_groups = chunks_df.groupby(
        "document",
        sort=False
    ).groups

    for document, indices in document_groups.items():

        indices = np.asarray(
            list(indices),
            dtype=np.int64
        )

        # ----------------------------------------------------
        # Get top semantic chunks in this PDF
        # ----------------------------------------------------

        semantic_values = semantic_scores[
            indices
        ]

        semantic_count = min(
            chunks_per_pdf,
            len(indices)
        )

        semantic_positions = np.argpartition(
            semantic_values,
            -semantic_count
        )[-semantic_count:]

        semantic_top = indices[
            semantic_positions
        ]

        # ----------------------------------------------------
        # Get top BM25 chunks in this PDF
        # ----------------------------------------------------

        bm25_values = bm25_scores[
            indices
        ]

        bm25_positions = np.argpartition(
            bm25_values,
            -semantic_count
        )[-semantic_count:]

        bm25_top = indices[
            bm25_positions
        ]

        # ----------------------------------------------------
        # Get top hybrid chunks in this PDF
        # ----------------------------------------------------

        hybrid_values = hybrid_scores[
            indices
        ]

        hybrid_positions = np.argpartition(
            hybrid_values,
            -semantic_count
        )[-semantic_count:]

        hybrid_top = indices[
            hybrid_positions
        ]

        # Combine all retrieval methods
        combined = set()

        combined.update(
            int(x) for x in semantic_top
        )

        combined.update(
            int(x) for x in bm25_top
        )

        combined.update(
            int(x) for x in hybrid_top
        )

        candidate_indices.extend(
            combined
        )

    # Remove duplicate chunk IDs
    candidate_indices = list(
        dict.fromkeys(candidate_indices)
    )

    print(
        f"✓ Candidate chunks: "
        f"{len(candidate_indices):,}"
    )

    # --------------------------------------------------------
    # 7. Rerank candidate chunks
    # --------------------------------------------------------

    pairs = []

    for idx in candidate_indices:

        pairs.append([
            query,
            chunks_df.iloc[idx]["text"]
        ])

    print(
        "\nRunning cross-encoder reranker..."
    )

    rerank_scores = reranker.predict(
        pairs,
        batch_size=rerank_batch_size,
        show_progress_bar=True
    )

    # --------------------------------------------------------
    # 8. Build results
    # --------------------------------------------------------

    results = []

    for idx, score in zip(
        candidate_indices,
        rerank_scores
    ):

        row = chunks_df.iloc[idx]

        results.append({

            "idx": int(idx),

            "document": row["document"],

            "page": int(row["page"]),

            "chunk_on_page": int(
                row["chunk_on_page"]
            ),

            "semantic_score": float(
                semantic_scores[idx]
            ),

            "bm25_score": float(
                bm25_scores[idx]
            ),

            "hybrid_score": float(
                hybrid_scores[idx]
            ),

            "rerank_score": float(score),

            "text": row["text"]

        })

    # --------------------------------------------------------
    # 9. Sort by reranker
    # --------------------------------------------------------

    results.sort(
        key=lambda x: x["rerank_score"],
        reverse=True
    )

    print(
        f"\n✓ Reranking complete"
    )

    # --------------------------------------------------------
    # 10. Statistics
    # --------------------------------------------------------

    documents_found = len(
        set(
            r["document"]
            for r in results
        )
    )

    print(
        f"✓ PDFs represented in candidates: "
        f"{documents_found:,}"
    )

    return results
```

```python
# ============================================================
# STEP 11 — COMPREHENSIVE EVIDENCE SELECTION
# ============================================================

def select_comprehensive_evidence(
    results,
    max_chunks_per_pdf=4,
    max_total_chunks=60,
    min_rerank_score=None
):
    """
    Select evidence across multiple PDFs.

    Each PDF can contribute multiple relevant chunks,
    while preventing one PDF from dominating.
    """

    selected = []

    document_counts = {}

    for result in results:

        document = result["document"]

        # Optional threshold
        if (
            min_rerank_score is not None
            and result["rerank_score"]
            < min_rerank_score
        ):
            continue

        current_count = document_counts.get(
            document,
            0
        )

        if current_count >= max_chunks_per_pdf:
            continue

        selected.append(result)

        document_counts[document] = (
            current_count + 1
        )

        if len(selected) >= max_total_chunks:
            break

    return selected
```

```python
# ============================================================
# STEP 12 — CITATION-SAFE COMPREHENSIVE CONTEXT
# ============================================================

def build_comprehensive_context(
    evidence,
    max_chars_per_chunk=5000
):
    """
    Build structured context for Qwen.

    Evidence is grouped by PDF so the model can understand
    which information came from which document.
    """

    if not evidence:
        return ""

    # --------------------------------------------------------
    # Group evidence by document
    # --------------------------------------------------------

    grouped = {}

    for item in evidence:

        document = item["document"]

        if document not in grouped:
            grouped[document] = []

        grouped[document].append(item)

    # --------------------------------------------------------
    # Build context
    # --------------------------------------------------------

    context_parts = []

    source_number = 1

    for document, items in grouped.items():

        # Sort by page
        items = sorted(
            items,
            key=lambda x: (
                x["page"],
                x["chunk_on_page"]
            )
        )

        document_block = []

        document_block.append(
            f"========== DOCUMENT {source_number} =========="
        )

        document_block.append(
            f"Document: {document}"
        )

        document_block.append(
            f"Evidence sections: {len(items)}"
        )

        document_block.append("")

        for j, item in enumerate(items, start=1):

            text = item["text"]

            if len(text) > max_chars_per_chunk:
                text = text[:max_chars_per_chunk]

            document_block.append(
                f"[EVIDENCE {source_number}.{j}]"
            )

            document_block.append(
                f"Document: {document}"
            )

            document_block.append(
                f"Page: {item['page']}"
            )

            document_block.append(
                f"Chunk: {item['chunk_on_page']}"
            )

            document_block.append(
                f"Reranker score: "
                f"{item['rerank_score']:.4f}"
            )

            document_block.append("")

            document_block.append(text)

            document_block.append(
                f"[/EVIDENCE {source_number}.{j}]"
            )

            document_block.append("")

        document_block.append(
            f"========== END DOCUMENT {source_number} =========="
        )

        context_parts.append(
            "\n".join(document_block)
        )

        source_number += 1

    return "\n\n".join(context_parts)


# ------------------------------------------------------------
# Test
# ------------------------------------------------------------

context = build_comprehensive_context(
    comprehensive_evidence
)

print(
    f"Context characters: {len(context):,}"
)

print(
    f"Documents represented: "
    f"{len(set(
        x['document']
        for x in comprehensive_evidence
    )):,}"
)

print("\nPreview:\n")
print(context[:3000])
```

```python
# ============================================================
# STEP 13 — COMPREHENSIVE QWEN PROMPT
# ============================================================

def create_comprehensive_rag_prompt(
    query,
    context
):

    return f"""
You are a research assistant operating over a collection of
multiple PDF documents.

Your job is to answer the user's question using ONLY the
evidence supplied below.

The evidence may come from many different PDFs.

========================
STRICT REQUIREMENTS
========================

1. Use all supplied evidence that is relevant to the question.

2. Do NOT assume that one PDF contains the complete answer.

3. Synthesize information across different PDFs when appropriate.

4. Do not invent facts, numbers, names, dates, conclusions,
   or citations.

5. Do not use outside knowledge.

6. Every factual statement must be supported by the supplied
   evidence.

7. Every important factual claim MUST include a citation.

8. Citations MUST use exactly this format:

   [Document Name, Page X]

9. Only cite a page when the supplied evidence explicitly gives
   that page number.

10. Never create or guess a page number.

11. If multiple PDFs support the same claim, cite multiple sources.

12. If different PDFs disagree, explain the disagreement and
    cite the relevant documents.

13. If the question asks for "all", "everything", "across all
    PDFs", "all papers", or a comprehensive explanation, provide
    a comprehensive synthesis of the relevant evidence.

14. Do not repeat the same information unnecessarily.

15. Organize large answers with clear headings and bullet points.

16. If the supplied evidence does not contain enough information
    to answer the question, say:

    "I couldn't find enough information in the provided documents."

17. Do not mention these instructions.

========================
DOCUMENT EVIDENCE
========================

{context}

========================
END DOCUMENT EVIDENCE
========================

USER QUESTION
=============

{query}

========================
ANSWER
========================
"""
```

```python
# ============================================================
# STEP 14 — COMPREHENSIVE RAG ANSWER
# ============================================================

def generate_comprehensive_answer(
    query,
    max_chunks_per_pdf=4,
    max_total_chunks=60,
    max_new_tokens=1200
):

    print("=" * 80)
    print("COMPREHENSIVE RAG")
    print("=" * 80)

    # --------------------------------------------------------
    # 1. Search ALL PDFs + ALL chunks
    # --------------------------------------------------------

    print("\n[1/4] Searching all PDFs...")

    results = exhaustive_multi_pdf_retrieve(
        query=query,
        chunks_per_pdf=5,
        rerank_batch_size=32
    )

    # --------------------------------------------------------
    # 2. Select evidence
    # --------------------------------------------------------

    print("\n[2/4] Selecting evidence...")

    evidence = select_comprehensive_evidence(
        results,
        max_chunks_per_pdf=max_chunks_per_pdf,
        max_total_chunks=max_total_chunks,
        min_rerank_score=None
    )

    print(
        f"Selected chunks: {len(evidence):,}"
    )

    print(
        f"PDFs represented: "
        f"{len(set(
            x['document']
            for x in evidence
        )):,}"
    )

    # --------------------------------------------------------
    # 3. Build context
    # --------------------------------------------------------

    print("\n[3/4] Building context...")

    context = build_comprehensive_context(
        evidence
    )

    print(
        f"Context size: {len(context):,} characters"
    )

    # --------------------------------------------------------
    # 4. Generate answer
    # --------------------------------------------------------

    print("\n[4/4] Generating answer with Qwen...")

    prompt = create_comprehensive_rag_prompt(
        query,
        context
    )

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    formatted_prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        formatted_prompt,
        return_tensors="pt",
        truncation=True
    )

    # Move tensors to GPU
    inputs = {
        key: value.to("cuda")
        for key, value in inputs.items()
    }

    # --------------------------------------------------------
    # Generate
    # --------------------------------------------------------

    with torch.inference_mode():

        output = llm.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=1.05,
            pad_token_id=tokenizer.eos_token_id
        )

    # --------------------------------------------------------
    # Remove prompt tokens
    # --------------------------------------------------------

    generated_tokens = output[
        0,
        inputs["input_ids"].shape[1]:
    ]

    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return {
        "answer": answer.strip(),
        "evidence": evidence,
        "num_pdfs": len(
            set(x["document"] for x in evidence)
        ),
        "num_chunks": len(evidence)
    }
```

```python
# ============================================================
# STEP 16 — TOKEN-AWARE CONTEXT MANAGEMENT
# ============================================================

def count_tokens(text):
    """
    Count tokens using the Qwen tokenizer.
    """
    return len(
        tokenizer.encode(
            text,
            add_special_tokens=False
        )
    )


def trim_text_to_tokens(
    text,
    max_tokens
):
    """
    Trim text without exceeding the token budget.
    """

    token_ids = tokenizer.encode(
        text,
        add_special_tokens=False
    )

    if len(token_ids) <= max_tokens:
        return text

    token_ids = token_ids[:max_tokens]

    return tokenizer.decode(
        token_ids,
        skip_special_tokens=True
    )


print("✓ Token utilities ready")
```

```python
# ============================================================
# STEP 17 — GROUP EVIDENCE BY PDF
# ============================================================

from collections import defaultdict


def group_evidence_by_pdf(evidence):

    grouped = defaultdict(list)

    for item in evidence:

        grouped[item["document"]].append(
            item
        )

    # Sort each PDF's evidence by page
    for document in grouped:

        grouped[document].sort(
            key=lambda x: (
                x["page"],
                x["chunk_on_page"]
            )
        )

    return dict(grouped)


# Test

grouped_evidence = group_evidence_by_pdf(
    comprehensive_evidence
)

print(
    f"✓ PDFs with evidence: "
    f"{len(grouped_evidence):,}"
)

for document, items in list(
    grouped_evidence.items()
)[:10]:

    print(
        f"{document}: "
        f"{len(items)} chunks"
    )
```

```python
# ============================================================
# STEP 18 — BUILD PDF EVIDENCE PACKAGES
# ============================================================

def build_pdf_evidence_package(
    document,
    evidence,
    max_tokens=3500
):
    """
    Build a token-limited evidence package for one PDF.
    """

    parts = []

    header = (
        f"DOCUMENT: {document}\n"
        f"IMPORTANT: Preserve the page numbers exactly.\n\n"
    )

    parts.append(header)

    current_tokens = count_tokens(
        header
    )

    for i, item in enumerate(
        evidence,
        start=1
    ):

        block = (
            f"[EVIDENCE {i}]\n"
            f"Page: {item['page']}\n"
            f"Chunk: {item['chunk_on_page']}\n"
            f"Text:\n"
            f"{item['text']}\n"
            f"[/EVIDENCE {i}]\n\n"
        )

        block_tokens = count_tokens(
            block
        )

        if current_tokens + block_tokens > max_tokens:

            remaining = (
                max_tokens - current_tokens
            )

            if remaining > 100:

                trimmed = trim_text_to_tokens(
                    block,
                    remaining
                )

                parts.append(trimmed)

            break

        parts.append(block)

        current_tokens += block_tokens

    return "".join(parts)


# Test

example_document = next(
    iter(grouped_evidence)
)

example_package = build_pdf_evidence_package(
    example_document,
    grouped_evidence[example_document]
)

print(example_package[:3000])

print(
    "\nTokens:",
    count_tokens(example_package)
)
```

```python
# ============================================================
# STEP 19 — MAP STAGE
# ============================================================

def create_map_prompt(
    query,
    document,
    evidence_package
):

    return f"""
You are analyzing ONE document as part of a larger
multi-document research task.

USER QUESTION:
{query}

DOCUMENT:
{document}

DOCUMENT EVIDENCE:
========================

{evidence_package}

========================
END DOCUMENT EVIDENCE

TASK:

Extract and summarize ONLY the information from this document
that is relevant to the user's question.

Requirements:

1. Use only the supplied evidence.
2. Do not use outside knowledge.
3. Preserve important facts, numbers, methods, findings,
   definitions, and conclusions.
4. Do not invent information.
5. Preserve exact page numbers.
6. Every factual statement must include:
   [Document Name, Page X]
7. If several pages support a point, cite all relevant pages.
8. Ignore irrelevant information.
9. Do not repeat the same fact unnecessarily.
10. If this document contains no useful information for the
    question, respond exactly:

    NO_RELEVANT_INFORMATION

DOCUMENT SUMMARY:
"""


def run_map_stage(
    query,
    grouped_evidence,
    max_input_tokens=3500,
    max_output_tokens=500
):

    summaries = []

    total_documents = len(
        grouped_evidence
    )

    print(
        f"MAP STAGE — {total_documents} PDFs"
    )

    for number, (
        document,
        evidence
    ) in enumerate(
        grouped_evidence.items(),
        start=1
    ):

        print(
            f"[{number}/{total_documents}] "
            f"{document}"
        )

        package = build_pdf_evidence_package(
            document,
            evidence,
            max_tokens=max_input_tokens
        )

        prompt = create_map_prompt(
            query,
            document,
            package
        )

        messages = [
            {
                "role": "user",
                "content": prompt
            }
        ]

        formatted_prompt = (
            tokenizer.apply_chat_template(
                messages,
                tokenize=False,
                add_generation_prompt=True
            )
        )

        inputs = tokenizer(
            formatted_prompt,
            return_tensors="pt",
            truncation=True,
            max_length=max_input_tokens
        )

        inputs = {
            key: value.to("cuda")
            for key, value in inputs.items()
        }

        with torch.inference_mode():

            output = llm.generate(
                **inputs,
                max_new_tokens=max_output_tokens,
                do_sample=False,
                repetition_penalty=1.05,
                pad_token_id=tokenizer.eos_token_id
            )

        generated_tokens = output[
            0,
            inputs["input_ids"].shape[1]:
        ]

        summary = tokenizer.decode(
            generated_tokens,
            skip_special_tokens=True
        ).strip()

        if summary and summary != "NO_RELEVANT_INFORMATION":

            summaries.append({
                "document": document,
                "summary": summary
            })

    print(
        f"\n✓ MAP COMPLETE"
    )

    print(
        f"Useful PDF summaries: "
        f"{len(summaries):,}"
    )

    return summaries
```

```python
# ============================================================
# STEP 21 — REDUCE STAGE
# ============================================================

def build_reduce_context(
    pdf_summaries,
    max_tokens=10000
):

    parts = []

    current_tokens = 0

    for item in pdf_summaries:

        block = (
            f"========== {item['document']} ==========\n"
            f"{item['summary']}\n"
            f"========== END {item['document']} ==========\n\n"
        )

        block_tokens = count_tokens(
            block
        )

        if current_tokens + block_tokens > max_tokens:
            break

        parts.append(block)

        current_tokens += block_tokens

    return "".join(parts)


def create_reduce_prompt(
    query,
    summaries_context
):

    return f"""
You are the final research synthesis assistant.

Answer the user's question using the supplied summaries from
MULTIPLE PDF documents.

USER QUESTION:
{query}

MULTI-DOCUMENT EVIDENCE:
==============================

{summaries_context}

==============================
END MULTI-DOCUMENT EVIDENCE

REQUIREMENTS:

1. Synthesize information across ALL supplied document summaries.

2. Do not assume that one document contains the complete answer.

3. Include relevant information from different documents.

4. Do not use outside knowledge.

5. Do not invent facts or citations.

6. Preserve important differences between documents.

7. If documents disagree, explicitly describe the disagreement.

8. Every important factual statement must have a citation.

9. Citations MUST use:

   [Document Name, Page X]

10. Preserve the exact document names and page numbers provided
    in the summaries.

11. Do not fabricate page numbers.

12. Avoid unnecessary repetition.

13. Organize the answer clearly.

14. If the question asks for comprehensive or all information,
    provide a broad synthesis of the supplied evidence.

15. If the evidence is insufficient, state:

    I couldn't find enough information in the provided documents.

FINAL ANSWER:
"""
```

```python
# ============================================================
# STEP 22 — FINAL MULTI-PDF SYNTHESIS
# ============================================================

def run_reduce_stage(
    query,
    pdf_summaries,
    max_context_tokens=10000,
    max_new_tokens=1500
):

    context = build_reduce_context(
        pdf_summaries,
        max_tokens=max_context_tokens
    )

    prompt = create_reduce_prompt(
        query,
        context
    )

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    formatted_prompt = (
        tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )
    )

    inputs = tokenizer(
        formatted_prompt,
        return_tensors="pt",
        truncation=True,
        max_length=max_context_tokens
    )

    inputs = {
        key: value.to("cuda")
        for key, value in inputs.items()
    }

    print(
        f"Reduce context tokens: "
        f"{inputs['input_ids'].shape[1]:,}"
    )

    with torch.inference_mode():

        output = llm.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=1.05,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_tokens = output[
        0,
        inputs["input_ids"].shape[1]:
    ]

    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    return answer
```

```python
# ============================================================
# STEP 23 — COMPLETE EXHAUSTIVE MULTI-PDF RAG
# ============================================================

def comprehensive_rag(
    query,
    chunks_per_pdf=5,
    max_chunks_per_pdf=4,
    max_total_evidence=100,
    map_input_tokens=3500,
    map_output_tokens=500,
    reduce_context_tokens=10000,
    reduce_output_tokens=1500
):

    print("\n")
    print("=" * 80)
    print("COMPREHENSIVE MULTI-PDF RAG")
    print("=" * 80)

    print(
        f"\nQuestion:\n{query}"
    )

    # ========================================================
    # PHASE 1 — SEARCH ALL PDFs / CHUNKS
    # ========================================================

    print("\n")
    print("[PHASE 1] EXHAUSTIVE RETRIEVAL")

    retrieval_results = (
        exhaustive_multi_pdf_retrieve(
            query=query,
            chunks_per_pdf=chunks_per_pdf,
            rerank_batch_size=32
        )
    )

    # ========================================================
    # PHASE 2 — SELECT EVIDENCE
    # ========================================================

    print("\n")
    print("[PHASE 2] EVIDENCE SELECTION")

    evidence = select_comprehensive_evidence(
        retrieval_results,
        max_chunks_per_pdf=max_chunks_per_pdf,
        max_total_chunks=max_total_evidence,
        min_rerank_score=None
    )

    print(
        f"Evidence chunks: "
        f"{len(evidence):,}"
    )

    print(
        f"PDFs represented: "
        f"{len(set(
            x['document']
            for x in evidence
        )):,}"
    )

    # ========================================================
    # PHASE 3 — GROUP BY PDF
    # ========================================================

    print("\n")
    print("[PHASE 3] GROUPING BY PDF")

    grouped = group_evidence_by_pdf(
        evidence
    )

    print(
        f"PDF groups: "
        f"{len(grouped):,}"
    )

    # ========================================================
    # PHASE 4 — MAP
    # ========================================================

    print("\n")
    print("[PHASE 4] MAP — ANALYZE EACH PDF")

    summaries = run_map_stage(
        query=query,
        grouped_evidence=grouped,
        max_input_tokens=map_input_tokens,
        max_output_tokens=map_output_tokens
    )

    # ========================================================
    # PHASE 5 — REDUCE
    # ========================================================

    print("\n")
    print("[PHASE 5] REDUCE — SYNTHESIZE")

    final_answer = run_reduce_stage(
        query=query,
        pdf_summaries=summaries,
        max_context_tokens=reduce_context_tokens,
        max_new_tokens=reduce_output_tokens
    )

    # ========================================================
    # RETURN
    # ========================================================

    return {
        "answer": final_answer,
        "evidence": evidence,
        "pdf_summaries": summaries,
        "pdfs_searched": chunks_df["document"].nunique(),
        "pdfs_with_evidence": len(grouped),
        "pdfs_summarized": len(summaries)
    }
```

```python
# ============================================================
# FIX — GRADIO SOURCE FORMATTERS
# ============================================================

import gradio as gr
import torch
import gc
from collections import defaultdict


def get_result_metadata(item):
    """
    Make retrieval results compatible with both:

    1. Normal RAG results
    2. Comprehensive RAG results
    """

    # --------------------------------------------------------
    # Comprehensive result already contains metadata
    # --------------------------------------------------------

    if "document" in item:

        return {
            "document": item["document"],
            "page": item["page"],
            "chunk_on_page": item.get(
                "chunk_on_page",
                0
            ),
            "text": item.get(
                "text",
                ""
            ),
            "rerank_score": item.get(
                "rerank_score",
                0.0
            )
        }

    # --------------------------------------------------------
    # Normal result only contains idx
    # --------------------------------------------------------

    idx = int(item["idx"])

    row = chunks_df.iloc[idx]

    return {
        "document": row["document"],
        "page": int(row["page"]),
        "chunk_on_page": int(
            row["chunk_on_page"]
        ),
        "text": row["text"],
        "rerank_score": float(
            item.get(
                "rerank_score",
                0.0
            )
        )
    }


# ============================================================
# SOURCE FORMATTER
# ============================================================

def format_pdf_sources(result):

    if not result:
        return "No retrieval information."

    evidence = result.get(
        "evidence",
        []
    )

    if not evidence:
        return "No relevant sources found."

    grouped = defaultdict(list)

    for item in evidence:

        metadata = get_result_metadata(item)

        grouped[
            metadata["document"]
        ].append(metadata)

    output = []

    output.append(
        "## 📚 Documents Used\n"
    )

    output.append(
        f"**PDFs searched:** "
        f"{result.get('pdfs_searched', 0)}"
    )

    output.append(
        f"\n**PDFs contributing evidence:** "
        f"{result.get('pdfs_with_evidence', 0)}"
    )

    if result.get("pdfs_summarized", 0) > 0:

        output.append(
            f"\n**PDFs summarized:** "
            f"{result['pdfs_summarized']}"
        )

    output.append("\n---")

    for document, items in grouped.items():

        pages = sorted(
            set(
                item["page"]
                for item in items
            )
        )

        page_text = ", ".join(
            str(page)
            for page in pages
        )

        output.append(
            f"\n### 📄 {document}"
        )

        output.append(
            f"Pages: **{page_text}**"
        )

        output.append(
            f"Evidence chunks: **{len(items)}**"
        )

    return "\n".join(output)


# ============================================================
# RETRIEVAL DETAILS
# ============================================================

def format_retrieval_details(result):

    if not result:
        return ""

    evidence = result.get(
        "evidence",
        []
    )

    if not evidence:
        return "No evidence retrieved."

    output = []

    output.append(
        "## 🔍 Retrieved Evidence"
    )

    for i, item in enumerate(
        evidence,
        start=1
    ):

        metadata = get_result_metadata(item)

        output.append(
            f"\n### {i}. "
            f"{metadata['document']} "
            f"— Page {metadata['page']}"
        )

        output.append(
            f"Reranker score: "
            f"`{metadata['rerank_score']:.4f}`"
        )

        preview = metadata["text"]

        if len(preview) > 800:
            preview = preview[:800] + "..."

        output.append(
            f"\n> {preview}"
        )

    return "\n".join(output)


print("✓ Gradio formatter fixed")
```

```python
# ============================================================
# CONVERSATIONAL RAG — QUERY REWRITING
# ============================================================

def rewrite_query_for_retrieval(
    current_question,
    history,
    max_history_messages=6
):
    """
    Convert a follow-up question into a standalone search query.

    Example:

    Previous:
        User: What methods were used for galaxy classification?
        AI: Several machine-learning methods...

    Current:
        Which one performed best?

    Rewritten:
        Which machine-learning method performed best for
        galaxy classification?
    """

    if not history:
        return current_question.strip()

    # --------------------------------------------------------
    # Keep only recent conversation
    # --------------------------------------------------------

    recent_history = history[
        -max_history_messages:
    ]

    history_text = []

    for message in recent_history:

        # Support Gradio message dictionaries
        if isinstance(message, dict):

            role = message.get(
                "role",
                ""
            )

            content = message.get(
                "content",
                ""
            )

            if content:

                history_text.append(
                    f"{role.upper()}: {content}"
                )

        # Support older tuple format
        elif isinstance(message, (list, tuple)):

            if len(message) >= 2:

                user_msg = message[0]
                assistant_msg = message[1]

                if user_msg:

                    history_text.append(
                        f"USER: {user_msg}"
                    )

                if assistant_msg:

                    history_text.append(
                        f"ASSISTANT: {assistant_msg}"
                    )

    history_text = "\n".join(
        history_text
    )

    prompt = f"""
You are a search-query rewriting assistant.

Your job is to rewrite the user's latest question into a
standalone search query for a document retrieval system.

CONVERSATION:
====================

{history_text}

====================

LATEST USER QUESTION:
{current_question}

RULES:

1. Preserve the user's actual intent.
2. Resolve references such as:
   "it", "they", "this", "that", "the first one",
   "the second method", etc.
3. Include important subject terminology from the conversation.
4. Do not answer the question.
5. Do not add facts that are not present in the conversation.
6. Keep the rewritten query concise.
7. Output ONLY the rewritten search query.

STANDALONE SEARCH QUERY:
"""

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    formatted_prompt = (
        tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )
    )

    inputs = tokenizer(
        formatted_prompt,
        return_tensors="pt",
        truncation=True,
        max_length=4000
    )

    inputs = {
        key: value.to("cuda")
        for key, value in inputs.items()
    }

    with torch.inference_mode():

        output = llm.generate(
            **inputs,
            max_new_tokens=100,
            do_sample=False,
            repetition_penalty=1.05,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_tokens = output[
        0,
        inputs["input_ids"].shape[1]:
    ]

    rewritten = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    # --------------------------------------------------------
    # Safety fallback
    # --------------------------------------------------------

    if not rewritten:

        return current_question.strip()

    return rewritten
```

```python
# ============================================================
# FIXED GRADIO INTERFACE
# COMPATIBLE WITH YOUR GRADIO VERSION
# ============================================================

import gradio as gr


with gr.Blocks(
    title="Multi-PDF Research Assistant"
) as demo:

    # ========================================================
    # HEADER
    # ========================================================

    gr.Markdown(
        """
# 📚 Multi-PDF Research Assistant

### Research across your entire PDF collection

**228 PDFs • BGE-M3 • FAISS • BM25 • Reranker • Qwen 14B • A100**

Ask questions about your documents and continue the conversation
with follow-up questions.
"""
    )

    # ========================================================
    # RETRIEVAL MODE
    # ========================================================

    mode = gr.Radio(
        choices=[
            "Normal",
            "Comprehensive"
        ],

        value="Normal",

        label="Retrieval Mode",

        info=(
            "Normal = focused retrieval. "
            "Comprehensive = search across the entire PDF collection."
        )
    )

    # ========================================================
    # CHAT
    # ========================================================

    chatbot = gr.Chatbot(
        label="Conversation",
        height=500
    )

    # ========================================================
    # USER INPUT
    # ========================================================

    message = gr.Textbox(
        label="Ask your documents",

        placeholder=(
            "Example: What do the papers say about galaxy morphology?"
        ),

        lines=3
    )

    with gr.Row():

        submit = gr.Button(
            "🔎 Search & Answer",
            variant="primary"
        )

        clear = gr.Button(
            "🗑️ Clear"
        )

    # ========================================================
    # ANSWER
    # ========================================================

    gr.Markdown(
        "## 🧠 Answer"
    )

    answer = gr.Markdown()

    # ========================================================
    # SOURCES
    # ========================================================

    sources = gr.Markdown()

    # ========================================================
    # RETRIEVAL DETAILS
    # ========================================================

    with gr.Accordion(
        "🔍 Retrieval Details",
        open=False
    ):

        details = gr.Markdown()

    # ========================================================
    # SUBMIT FUNCTION
    # ========================================================

    def submit_message(
        message,
        history,
        mode
    ):

        if not message or not message.strip():

            return (
                history,
                "",
                "",
                "",
                ""
            )

        if history is None:

            history = []

        # ----------------------------------------------------
        # Convert old Gradio history into our expected format
        # ----------------------------------------------------

        history_for_rag = []

        for item in history:

            if isinstance(item, (list, tuple)):

                if len(item) >= 2:

                    user_message = item[0]
                    assistant_message = item[1]

                    if user_message:

                        history_for_rag.append({
                            "role": "user",
                            "content": user_message
                        })

                    if assistant_message:

                        history_for_rag.append({
                            "role": "assistant",
                            "content": assistant_message
                        })

        # ----------------------------------------------------
        # Run RAG
        # ----------------------------------------------------

        answer_text, sources_text, details_text = (
            gradio_rag(
                message,
                history_for_rag,
                mode
            )
        )

        # ----------------------------------------------------
        # Add conversation turn
        # ----------------------------------------------------

        new_history = history + [
            (
                message,
                answer_text
            )
        ]

        return (
            new_history,
            answer_text,
            sources_text,
            details_text,
            ""
        )

    # ========================================================
    # BUTTON
    # ========================================================

    submit.click(

        submit_message,

        inputs=[
            message,
            chatbot,
            mode
        ],

        outputs=[
            chatbot,
            answer,
            sources,
            details,
            message
        ]
    )

    # ========================================================
    # ENTER KEY
    # ========================================================

    message.submit(

        submit_message,

        inputs=[
            message,
            chatbot,
            mode
        ],

        outputs=[
            chatbot,
            answer,
            sources,
            details,
            message
        ]
    )

    # ========================================================
    # CLEAR
    # ========================================================

    def clear_chat():

        return (
            [],
            "",
            "",
            ""
        )

    clear.click(

        clear_chat,

        outputs=[
            chatbot,
            answer,
            sources,
            details
        ]
    )


print("✓ Gradio interface created successfully")
```
